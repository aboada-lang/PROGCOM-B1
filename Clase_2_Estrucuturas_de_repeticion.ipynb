# Estructura de repeticion
se usa para repetir acciones

```
num=int(input("Ingresar numero"))
for i in range(1,11):
  print(f"{num} x {i} = {num*i}")
```

```
n=int(input("Ingresar el valor de n: "))
suma = 0
for i in range(1,n,1):
  if i % 2 ==0:
    suma-= 1/i
    print(f"-1{i}", end=" ")
else:
  suma +=1 / i
  print(f"+ 1/{i}" if i > 1 else "1", end= " ")
print("resultado:", suma)
```

##estrucutura mientras (while)
Estructura computacional de repeticion, empleada para repetir una accion **Mientras**

```
for par in range (2,21,2):
  print(par)
```

```
C=input("Ingrese su parte de la canción: ")
if C.lower()== "quién es?" or C.lower()=="¿quién es?" or C.lower()=="quien es?":
  C1=1
elif C.lower() == "que vienes a buscar?" or C.lower()== "qué vienes a buscar?" or C.lower()=="¿que vienes a buscar?":
  C1=2
while C1== 1:
  print("Soy yo.")
  break

while C1== 2:
  print("A ti.")
  break
```

```
experiencias = []

# Pedir la experiencia de los 28 estudiantes
for i in range(28):
    experiencia = float(input("Ingrese la experiencia del estudiante " + str(i + 1) + ": "))
    experiencias.append(experiencia)

# Calcular el promedio
promedio = sum(experiencias) / 28

print("El promedio de experiencia es:", promedio)

# Mostrar los que tienen más experiencia
print("Estudiantes con más experiencia que el promedio:")

for i in range(28):
    if experiencias[i] > promedio:
        print("Estudiante", i + 1, ":", experiencias[i], "años")
```
